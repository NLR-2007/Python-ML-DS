Bitwise-Operators--> Works on Binary values.

```python
# Bitwise Operators in Python

a = 5      # Binary: 0101
b = 3      # Binary: 0011

print("a =", a)
print("b =", b)

# 1. Bitwise AND (&)
print("\n1. AND (&)")
print("a & b =", a & b)     # 0101 & 0011 = 0001 = 1

# 2. Bitwise OR (|)
print("\n2. OR (|)")
print("a | b =", a | b)     # 0101 | 0011 = 0111 = 7

# 3. Bitwise XOR (^)
print("\n3. XOR (^)")
print("a ^ b =", a ^ b)     # 0101 ^ 0011 = 0110 = 6

# 4. Bitwise NOT (~)
print("\n4. NOT (~)")
print("~a =", ~a)

# 5. Left Shift (<<)
print("\n5. Left Shift (<<)")
print("a << 1 =", a << 1)   # 0101 -> 1010 = 10

# 6. Right Shift (>>)
print("\n6. Right Shift (>>)")
print("a >> 1 =", a >> 1)   # 0101 -> 0010 = 2
```

<b>Lets Build the "Guessing Game"</b>

```python
import random
jackpot = random.randint(1,100)
print("NLR--Number Guessing Game(NGG)")
guess = int(input("Guess the number: "))
count = 1;
while guess!=jackpot:
  if guess<jackpot:
    print("Guess Higher")
  else:
    print("Guess Lower")
  guess = int(input("Guess the number: "))
  count +=1
print("Correct. You gussed right one.")
print("You gussed in",count,"no of steps.")
```
