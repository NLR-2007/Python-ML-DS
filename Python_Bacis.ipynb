<b>Identifiers</b>
Are used to identify the varables, functions, class, module, or the object.<br>
Rules for Identifiers in Python<br>
1. Identifier names must start with a letter (A-Z or a-z) or underscore (_).<br>
2. They cannot start with a number.
3. An identifier can contain:
letters<br>
digits<br>
underscores (_)<br>
4. Special symbols like @, #, $, %, - are not allowed.<br>
5. Spaces are not allowed in identifiers.<br>
6. Python keywords cannot be used as identifiers.<br>

7. Python identifiers are case-sensitive.<br>
Example:<br>

age<br>
Age<br>

Both are different.<br>
Note: Keywords cannot be used as Identifiers
